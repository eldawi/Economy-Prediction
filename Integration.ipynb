```python
import geopandas as gpd

# Load VIIRS GeoDataFrame
viirs_gdf = gpd.read_file('maldives_viirs_combined.shp')

# Load Ookla GeoDataFrame
ookla_gdf = gpd.read_file('maldives_ookla_tiles.shp')
```

```python
# Check CRS
print(viirs_gdf.crs)
print(ookla_gdf.crs)

# Reproject Ookla to match VIIRS CRS, if needed
ookla_gdf = ookla_gdf.to_crs(viirs_gdf.crs)
```

```python
print(viirs_gdf.columns)
print(ookla_gdf.columns)
```

```python
gdp_values = {2021: 55.3, 2022: 2.3, 2023: 4.0}  # Example GDP values per year
viirs_gdf['GDP_at_Market_Price'] = viirs_gdf['year'].map(gdp_values)

```

```python
viirs_gdf.rename(columns={'light_in_1': 'annual radiance'}, inplace=True)

viirs_gdf.rename(columns={'light_inte': 'light intensity'}, inplace=True)
```

```python
print(viirs_gdf.head())
```

```python
unified_gdf = gpd.sjoin(ookla_gdf, viirs_gdf, how='inner', predicate='intersects')
```

```python
unified_gdf.to_csv('unified_dataset.csv', index=False)
unified_gdf.to_file('unified_dataset.shp')
```
