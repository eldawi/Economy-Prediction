```python
import geopandas as gpd
import pandas as pd

# Load the Ookla SHP files for 2021, 2022, and 2023
ookla_2021 = gpd.read_file('maldives_ookla_tiles_2021.shp')
ookla_2022 = gpd.read_file('maldives_ookla_tiles_2022.shp')
ookla_2023 = gpd.read_file('maldives_ookla_tiles.shp')

# Add a 'year' column to each dataset
ookla_2021['year'] = 2021
ookla_2022['year'] = 2022
ookla_2023['year'] = 2023

# Combine all years into one GeoDataFrame
ookla_data = gpd.GeoDataFrame(pd.concat([ookla_2021, ookla_2022, ookla_2023], ignore_index=True))

print(ookla_data.head())
```

```python
# Group by year and calculate annual averages for download speed
annual_avg_speeds = ookla_data.groupby('year', as_index=False)['avg_d_kbps'].mean()

# Merge the annual averages back with the original GeoDataFrame (to keep spatial info)
ookla_data = ookla_data.merge(annual_avg_speeds, on='year', suffixes=('', '_annual'))

print("Aggregated GeoDataFrame:")
print(ookla_data.head())
```

```python
ookla_data.to_file('maldives_ookla_aggregated.shp')
print("Aggregated SHP file saved as 'maldives_ookla_aggregated.shp'")
```

```python
# Drop the geometry column before saving as CSV
ookla_data.drop(columns='geometry').to_csv('maldives_ookla_aggregated.csv', index=False)
print("Aggregated CSV file saved as 'maldives_ookla_aggregated.csv'")
```
